```python
df.head()
```

```python
col_to_split = 'Type of Breach'
col_to_split1 = 'Location of Breached Information'

df[col_to_split] = df[col_to_split].astype(str).str.split(', ')
df[col_to_split1] = df[col_to_split1].astype(str).str.split(', ')

df_exploded = df.explode(col_to_split).explode(col_to_split1).reset_index(drop=True)
df_exploded.head()
```

```python
frekuensi = df_exploded["Covered Entity Type"].value_counts()
persentase = df_exploded["Covered Entity Type"].value_counts(normalize=True) * 100

# Digabung jadi satu tabel biar rapih
ringkasan = pd.DataFrame({
    "Frekuensi": frekuensi,
    "Persentase (%)": persentase.round(2)
})
print(ringkasan)

modus = df_exploded["Covered Entity Type"].mode()[0]
print(f"\nData yang paling sering muncul: {modus}")
```
